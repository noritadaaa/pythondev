```python
class Baseball_Team:
  def __init__(self, name, win, lose, draw): #コンストラクタ
    self.name = name #インスタンスメンバ変数
    self.win = win
    self.lose = lose
    self.draw = draw

  def calc_win_rate(self): #勝率の計算メソッド
    return self.win / (self.win + self.lose) #メンバ変数で持つのではなく、メソッド実行の都度計算して値を返す

  def show_team_result(self): #結果表示のメソッド
    win_rate = self.calc_win_rate() #計算結果を変数に一旦入れる
    print(f"{self.name:<10} {self.win:>4} {self.lose:>4} {self.draw:>4} {win_rate:>6.3f}")
```

```python
teams = [
  Baseball_Team("Archers", 77, 64, 2),
  Baseball_Team("Meteors", 71, 69, 3),
  Baseball_Team("Cheetahs", 69, 68, 6),
  Baseball_Team("Sharks", 70, 70, 3),
  Baseball_Team("Pythons", 68, 73, 2),
  Baseball_Team("Owls", 59, 82, 2)
] #各球団のインスタンスを生成してリストに格納

print(f"{'team':<10} {'win':>4} {'lose':>4} {'draw':>4} {'rate':>4}")
for team in teams: team.calc_win_rate() #各球団毎に勝率の計算
for team in teams: team.show_team_result() #各球団毎に結果を表示
```
