```python
# テストの平均点を求める
scores = [] # 入力された点数をそれぞれ格納（list型）
count = 0 # 入力した人数のカウンタ変数
while True:
  score = int(input("点数を入力してください："))
  if score == -1:
    break
  scores.append(score) # 入力されたscoreをscoresの最後に格納
  count += 1
  score_total = sum(scores) # scoresの値を合計してscore_totalに代入
  score_avg = score_total / count
  score_avg = round(score_avg, 2)
print(f"{count}人のテストの平均点は{score_avg}点です")
```
