````python
# 内包表記でFizzBuzz判定
result = [
  "FizzBuzz" if n % 3 == 0 and n % 5 ==0 else
  "Fizz" if n % 3 == 0 else
  "Buzz" if n % 5 == 0 else
  str(n)
  for n in range(1, 101)
]

print(result)
````

````python
#内包表記でFizzBuzz判定(関数利用)
def fizz_buzz(n):
  if n % 15 == 0:
    return "FizzBuzz"
  else:
    if n % 3 == 0:
      return "Fizz"
    else:
      if n % 5 == 0:
        return "Buzz"
      else:
        return str(n)

result = [fizz_buzz(n) for n in range(1, 101)]
print(result)
````

普通のif文と、内包表記での式のifでは記述が微妙に異なる

■if文（上から下へ読む、フローを意識する、最終的に処理結果を求めている）
```
if 条件:
  A
else:
  B
```

※「else」は return / break / continue / raise がある時は省略したほうが可読性いいかも？
※「else」は残り全部を明示する（2択ならあり）
else省略版
```
def fizz_buzz(n):
  if n % 15 == 0:
    return "FizzBuzz"
  if n % 3 == 0:
    return "Fizz"
  if n % 5 == 0:
    return "Buzz"
  return str(n)
```

■内包表記の条件式（計算式と同じ感覚、最終的に値を決定する計算なので）
```
A if 条件 else B
```
A ← 条件1
B ← 条件2
