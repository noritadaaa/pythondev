```python
words = []
alphabet = dict.fromkeys(list("abcdefghijklmnopqrstuvwxyz"), 0)

while True:
  word = input("英単語を入力してください：").lower().strip() #スペースは未入力扱い
  if not word: #未入力の場合はループ抜ける
    break
  for ch in word: #入力された単語を一文字ずつ格納
    if ch in alphabet.keys(): #alphabetの中に入力された一文字があれば
      alphabet[ch] += 1 #alphabetから該当の値を取り出して+1して格納する
  words.append(word)

words.sort() #wordsに格納されている文字をABC順（昇順）で出力する
print(f"入力した単語：{words}")

for key, value in alphabet.items():
  if value > 0: #alphabetの中で値が1以上のもの
    print(f"{key}が{value}個ありました") #キーと値セットで出力する
```
