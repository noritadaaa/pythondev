```python
#1~100までの中の素数を表示する
for i in range(1, 101):
  if i != 1: #1ではない場合進む
    if i == 2 or i % 2 != 0: #2は進む。もしくは2で割って余り0（2の倍数）ではない場合進む
      if i == 3 or i % 3 != 0: #3は進む。もしくは3で割って余り0（3の倍数）ではない場合進む
        if i == 5 or i % 5 != 0: #5は進む。もしくは5で割って余り0（5の倍数）ではない場合進む
          if i == 7 or i % 7 != 0: #7は進む。もしくは7で割って余り0（7の倍数）ではない場合進む
            print(i)
```
