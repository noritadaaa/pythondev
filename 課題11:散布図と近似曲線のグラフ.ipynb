```python
import numpy as np

x = np.array([62, 95, 83, 100, 72, 71, 55, 85, 67, 75])   # english
y = np.array([62, 80, 85, 95, 58, 76, 52, 84, 67, 71])    # math
```

```python
import matplotlib.pyplot as plt
p = np.polyfit(x, y, 1)
f = np.poly1d(p)

plt.xlabel("english")
plt.ylabel("math")
plt.scatter(x, y, label="score")
plt.plot(x, f(x), color="Red", label="Lsm") #最小二乗法(Least Squares Method)の1次関数
plt.legend() #ラベルの表示に必要
plt.show()
```
